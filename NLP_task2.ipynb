```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve LookupError
text = "tokenization without transformation is straightforward with tools like NLTK"
tokens = word_tokenize(text)
print("tokens:",tokens)
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
text = "tokenization without transformation is straightforward with tools like NLTK"
tokens_transformer = tokenizer(text,return_tensors="pt")
print("transformers token:",tokens_transformer)
tokens_transformers_list = tokenizer.convert_ids_to_tokens(tokens_transformer['input_ids'][0].numpy().tolist())
print("transformers tokens(list):",tokens_transformers_list)
decoded_text = tokenizer.decode(tokens_transformer['input_ids'][0],skip_special_tokens = True)
print("Decoded text:",decoded_text)
```
